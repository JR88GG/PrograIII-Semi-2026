```python
def suma(num1,num2):
    return float(num1) + float(num2)
```

```python
def resta(num1,num2):
    return float(num1) - float(num2)
```

Hacer el siguiente ejercicio en python haciendo uso de funciones, que conviertan grados celcius a  ferenheit


```python
def convertirC(celcius):
  return float(celcius)*9/5 +32
```

```python
c = input("Grados Celcius: ")
print("Grados Ferenheit: ", convertirC(c))
```

Hacer el siguiente ejercicio en python haciendo uso de funciones, que conviertan grados ferenheit a celcius

```python
def convertirf(F):
  return (float(F)-32)*5/9
```

```python
f = input("Grados ferenheit: ")
print("Grados Celcius: ", convertirf(f))
```

Hacer el siguiente ejercicio en python haciendo uso de funciones, que conviertan grados celcius a kelvin

```python
def convertirK(c):
  return float(c)+273.15
```

```python
c = input("Grados Celcius: ")
print("Grados Kelvin: ", convertirK(c))
```

Hacer el siguiente ejercicio en python haciendo uso de funciones anonimas, que conviertan grados ferenheit a celcius

```python
convertir = lambda c: float(c)*9/5+32
```

```python
c = float(input("Grados Celcius: "))
print("Grados Ferenheit: ", convertir(c))
```

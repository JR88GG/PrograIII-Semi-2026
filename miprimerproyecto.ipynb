```python
#calcular sueldo
# Tabla del ISR
tabla_isr = [
    [0.01, 550, 0, 0],
    [550.01, 895.24, 0.10, 17.67],
    [895.25, 2038.10, 0.20, 60],
    [2038.11, 9999999, 0.30, 288.57]
]

# Funciones lambda
calcular_sueldo = lambda horas, valor_hora: horas * valor_hora * 30

calcular_deduccion = lambda sueldo, porcentaje: sueldo * porcentaje

calcular_formula_isr = lambda sueldo, limite, porcentaje, cuota: \
    (sueldo - limite) * porcentaje + cuota


# Función para calcular ISR
def calcular_isr(sueldo):
    isr = 0

    for fila in tabla_isr:
        limite_inferior = fila[0]
        limite_superior = fila[1]
        porcentaje = fila[2]
        cuota = fila[3]

        if limite_inferior <= sueldo <= limite_superior:
            isr = calcular_formula_isr(
                sueldo,
                limite_inferior,
                porcentaje,
                cuota
            )

    return isr


# Entrada de datos
horas = float(input("Ingrese las horas trabajadas por día: "))
valor_hora = float(input("Ingrese el valor de la hora: $"))


# Calcular sueldo mensual
sueldo = calcular_sueldo(horas, valor_hora)


# Calcular ISSS y AFP
isss = calcular_deduccion(sueldo, 0.03)
afp = calcular_deduccion(sueldo, 0.0725)


# Calcular ISR sobre el sueldo después de ISSS y AFP
sueldo_para_isr = sueldo - isss - afp
isr = calcular_isr(sueldo_para_isr)


# Calcular total de deducciones y sueldo neto
total_deducciones = isss + afp + isr
sueldo_neto = sueldo - total_deducciones


# Mostrar resultados
print("\n========== RESULTADOS ==========")
print(f"Horas trabajadas por día: {horas:.2f}")
print(f"Valor de la hora:         ${valor_hora:.2f}")
print(f"Sueldo mensual:           ${sueldo:.2f}")

print("\n------- DEDUCCIONES -------")
print(f"ISSS (3%):                ${isss:.2f}")
print(f"AFP (7.25%):              ${afp:.2f}")
print(f"ISR:                      ${isr:.2f}")
print(f"Total deducciones:        ${total_deducciones:.2f}")

print("\nSueldo neto:               ${:.2f}".format(sueldo_neto))
```

```python
#Escribir funcion para calcular nota.
# Function to calculate the final note for a subject
def calculate_subject_final_note(subject_notas):
    c1 = subject_notas["c1"]["lab1"] * 0.30 + \
         subject_notas["c1"]["lab2"] * 0.30 + \
         subject_notas["c1"]["parcial"] * 0.40

    c2 = subject_notas["c2"]["lab1"] * 0.30 + \
         subject_notas["c2"]["lab2"] * 0.30 + \
         subject_notas["c2"]["parcial"] * 0.40

    c3 = subject_notas["c3"]["lab1"] * 0.30 + \
         subject_notas["c3"]["lab2"] * 0.30 + \
         subject_notas["c3"]["parcial"] * 0.40

    final_note = (c1 + c2 + c3) / 3
    return final_note
```

```python
from functools import reduce
class estadistica:
  def media(self, serie):
    suma = reduce(lambda acum, num: acum+ float(num), serie, 0)
    return suma/len(serie)

  def varianza(self, serie):
    media = self.media(serie)
    #4 + 0 + 4 + 0 + 0 + 16= 24/6 = 4.0
    suma = reduce(lambda acum, num: acum + (float(num)-media)**2, serie, 0)
    return suma/len(serie)

  def desviacion(self, serie):
    return self.varianza(serie)**0.5
```

```python
#Hacer un programa en python que lea una serie de numeros introducida por el usuario
#y que calcule la media aritmetica
objEstadistica = estadistica()
serie = input("Serie: ").split(",")
print("Media: ", objEstadistica.media(serie))
print("Varianza: ", objEstadistica.varianza(serie))
print("Desviacion: ", objEstadistica.desviacion(serie))
```

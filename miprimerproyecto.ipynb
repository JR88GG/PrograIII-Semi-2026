```python
#calcular sueldo
# Tabla del ISR
tabla_isr = [
    [0.01, 550, 0, 0],
    [550.01, 895.24, 0.10, 17.67],
    [895.25, 2038.10, 0.20, 60],
    [2038.11, 9999999, 0.30, 288.57]
]

# Funciones lambda
calcular_sueldo = lambda horas, valor_hora: horas * valor_hora * 30

calcular_deduccion = lambda sueldo, porcentaje: sueldo * porcentaje

calcular_formula_isr = lambda sueldo, limite, porcentaje, cuota: \
    (sueldo - limite) * porcentaje + cuota


# Función para calcular ISR
def calcular_isr(sueldo):
    isr = 0

    for fila in tabla_isr:
        limite_inferior = fila[0]
        limite_superior = fila[1]
        porcentaje = fila[2]
        cuota = fila[3]

        if limite_inferior <= sueldo <= limite_superior:
            isr = calcular_formula_isr(
                sueldo,
                limite_inferior,
                porcentaje,
                cuota
            )

    return isr


# Entrada de datos
horas = float(input("Ingrese las horas trabajadas por día: "))
valor_hora = float(input("Ingrese el valor de la hora: $"))


# Calcular sueldo mensual
sueldo = calcular_sueldo(horas, valor_hora)


# Calcular ISSS y AFP
isss = calcular_deduccion(sueldo, 0.03)
afp = calcular_deduccion(sueldo, 0.0725)


# Calcular ISR sobre el sueldo después de ISSS y AFP
sueldo_para_isr = sueldo - isss - afp
isr = calcular_isr(sueldo_para_isr)


# Calcular total de deducciones y sueldo neto
total_deducciones = isss + afp + isr
sueldo_neto = sueldo - total_deducciones


# Mostrar resultados
print("\n========== RESULTADOS ==========")
print(f"Horas trabajadas por día: {horas:.2f}")
print(f"Valor de la hora:         ${valor_hora:.2f}")
print(f"Sueldo mensual:           ${sueldo:.2f}")

print("\n------- DEDUCCIONES -------")
print(f"ISSS (3%):                ${isss:.2f}")
print(f"AFP (7.25%):              ${afp:.2f}")
print(f"ISR:                      ${isr:.2f}")
print(f"Total deducciones:        ${total_deducciones:.2f}")

print("\nSueldo neto:               ${:.2f}".format(sueldo_neto))
```

TAREA

1-Econtrar la media armonica

2-Encontrar la frecuencia

3-Escriba un programa en python que determine si es una palabra introducida por el usuario es palindrome. Ej: ana, oso, oro, reconocer, bob, somos, salas, ala, etc. Haciendo uso de clases, objetos, funciones, etc

```python
"""
==========================================================
 TAREA: MEDIA ARMONICA, FRECUENCIA Y PALINDROMOS
==========================================================
1- Media armonica
2- Frecuencia
3- Verificador de palindromos (usando clases y objetos)
==========================================================
"""

from collections import Counter
import unicodedata


# ----------------------------------------------------------------
# CLASE 1: ESTADISTICA (media armonica y frecuencia)
# ----------------------------------------------------------------
class Estadistica:
    """Calcula la media armonica y la frecuencia de una lista de numeros."""

    def __init__(self, datos):
        if not datos:
            raise ValueError("La lista de datos no puede estar vacia.")
        if any(x == 0 for x in datos):
            raise ValueError("La media armonica no se puede calcular con ceros en los datos.")
        self.datos = list(datos)

    def media_armonica(self):
        """
        La media armonica es el numero de datos (n) dividido entre la
        suma de los inversos (1/x) de cada dato. Se usa mucho cuando
        los datos son razones o velocidades (ej: promedio de velocidades).
        Formula: n / (1/x1 + 1/x2 + ... + 1/xn)
        """
        n = len(self.datos)
        suma_inversos = sum(1 / x for x in self.datos)
        return n / suma_inversos

    def frecuencia(self):
        """Diccionario {valor: cuantas veces aparece} ordenado por valor."""
        return dict(sorted(Counter(self.datos).items()))


# ----------------------------------------------------------------
# CLASE 2: PALINDROMO
# ----------------------------------------------------------------
class Palabra:
    """Representa una palabra y verifica si es palindromo."""

    def __init__(self, texto):
        self.original = texto
        self.normalizada = self._normalizar(texto)

    @staticmethod
    def _normalizar(texto):
        """
        Deja la palabra en minusculas, le quita tildes/acentos y elimina
        espacios, para que "Ana", "ana " o "Sofía" se comparen bien.
        """
        texto = texto.strip().lower()
        texto = texto.replace(" ", "")
        # Quita acentos (á -> a, é -> e, etc.)
        texto = "".join(
            c for c in unicodedata.normalize("NFD", texto)
            if unicodedata.category(c) != "Mn"
        )
        return texto

    def es_palindromo(self):
        """Una palabra es palindromo si se lee igual al derecho y al reves."""
        return self.normalizada == self.normalizada[::-1]

    def __str__(self):
        return self.original


# ----------------------------------------------------------------
# FUNCIONES DE USUARIO
# ----------------------------------------------------------------
def pedir_datos_numericos():
    while True:
        texto = input("Ingrese los datos separados por comas (ej: 4,8,6,5,3): ")
        try:
            datos = [float(x.strip()) for x in texto.split(",") if x.strip() != ""]
            if not datos:
                raise ValueError
            return datos
        except ValueError:
            print("Entrada invalida. Escriba solo numeros separados por comas.\n")


def verificar_palindromo():
    palabra_texto = input("Escriba una palabra: ")
    palabra = Palabra(palabra_texto)  # se crea un objeto de la clase Palabra
    if palabra.es_palindromo():
        print(f'\n"{palabra}" SI es un palindromo.\n')
    else:
        print(f'\n"{palabra}" NO es un palindromo.\n')


# ----------------------------------------------------------------
# MENU PRINCIPAL
# ----------------------------------------------------------------
def menu_principal():
    while True:
        print("\n==============================================")
        print(" TAREA: MEDIA ARMONICA / FRECUENCIA / PALINDROMOS")
        print("==============================================")
        print("1-Media armonica")
        print("2-Frecuencia")
        print("3-Verificar si una palabra es palindromo")
        print("0-Salir")

        opcion = input("Seleccione una opcion: ").strip()

        match opcion:
            case "1":
                datos = pedir_datos_numericos()
                estadistica = Estadistica(datos)
                print(f"\nMedia armonica: {round(estadistica.media_armonica(), 2)}\n")
            case "2":
                datos = pedir_datos_numericos()
                estadistica = Estadistica(datos)
                print(f"\nFrecuencia: {estadistica.frecuencia()}\n")
            case "3":
                verificar_palindromo()
            case "0":
                print("Fin del programa. ¡Hasta luego!")
                break
            case _:
                print("Opcion no valida.")


if __name__ == "__main__":
    menu_principal()
```
